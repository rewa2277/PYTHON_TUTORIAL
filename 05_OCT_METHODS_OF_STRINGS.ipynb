STRING METHODS:

```python
#METHODS : FUNCTIONS THAT ARE RELATED TO OBJECTS.
```

upper() : converts whole string into capital letters.


```python
str = "good morning , i am rewa your python educator."
str.upper()
```

lower() : converts whole string into small letters.

```python
str = "I AM ATTENDING LECTURE THAT CONDUCT BY SEVEN MENTOR"
str.lower()
```

title(): converts first letter of each character into capital letter.

```python
str = "I AM ATTENDING LECTURE THAT CONDUCT BY SEVEN MENTOR"
str.title()
```

capitalize() : convert first letter of first character.

```python
str = "I AM ATTENDING LECTURE THAT CONDUCT BY SEVEN MENTOR"
str.capitalize()
```

swapcase() : converts capital to small and vice versa

```python
str = "I AM ATTENDING LECTURE THAT CONDUCT BY SEVEN MENTOR"
str.swapcase()

```

str = "good morning , i am rewa your python educator."

find(): give particular letter index

```python
str = "Good Morning ,  I am your teacher"
str.find("t")
```

```python
str.find("z")  # gives -1 if substring is nor present
```

count : counts the occurance of the substring in string

```python
text = "banana"
print(text.count("a"))
```

startswith() : checks whether the string is starts with python

```python
text = "python is my favourite subject"
print(text.startswith("python"))
print(text.startswith("p"))
print(text.startswith("Python"))
```

endswith() : checks whether string end with or not

```python
text = "Python Programming"
print(text.endswith("Programming"))
print(text.endswith("P"))
print(text.endswith("g"))
```

replace() = it is used to replace the substring

```python
# syntax = .replace("old_string","new_string")
```

```python
text = "I like python"
print(text.replace("python","java"))
```

```python
text = "I like Python. Python is interesting."
print(text.replace("Python","Java"))
```
