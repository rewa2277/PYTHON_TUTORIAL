```python
char = "Python"
print(char[:4])
print(char[3:])
print(char[:])
print(char[-1:])
print(char[:-1])
```

```python
# HW : take a string from user and check whether it is palindrome or not
```

```python
text = input("Enter any string:")
reversed_str = text[::-1] 
if text == reversed_str :
    #palindrome = madam , civic , racecar
    print("It is a palindrome.")
else:
    print("It is not a palindrome.")
    
```
