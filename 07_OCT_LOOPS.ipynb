```python
list = [20,40,56,87,13,34]
largest = list[0]
for num in list:
    if num > largest :
        largest = num
print("Largest number is:",largest)
```

```python
list = [ 20,30,40,75,1 ]
smallest = list[0]
for num in list:
    if num < smallest:
        smallest = num
print("Smallest number is : ",smallest)
```
