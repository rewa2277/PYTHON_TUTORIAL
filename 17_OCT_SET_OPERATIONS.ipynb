```python
#TASK : 
#Solve using set operations:
website_customers = {101,102,103,104,105}
app_customers = {103,104,105,106,107}
```

```python
#questions:
# find the  common customers : 
common_customers = website_customers & app_customers 
print(common_customers)
```

```python
# find the total unique customers 
unique_customers =  website_customers | app_customers
print(unique_customers)
```

```python
# find only website customers 
only_website = website_customers - app_customers
print(only_website)
```

```python
# find only app customers 
only_app = app_customers-website_customers
print(only_app)
```

```python
#find the customers who use any one but does not uses both
common_remove =  website_customers ^ app_customers
print(common_remove)
```

SET COMPREHENSION: 

```python
#find salaries greater than 20000
salaries = {10000,25000,35000,40000}  
higher_sal = {i for i in salaries if i>20000}
print(higher_sal)
```

```python
set1 = {1,2,3,4}
#find the square of each element using set comprehension
square = {i**2 for i in set1}
print(square)
```

```python
#find unique even numbers in the below list using set comprehension
```

```python
data = {10,20,20,31,31,42,50,50}
unique_even = {i for i in data if i%2==0}
print(unique_even)
```
