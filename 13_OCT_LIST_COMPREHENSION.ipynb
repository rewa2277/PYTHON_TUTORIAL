LIST COMPREHENSION:

traditional method of squaring number:

```python
num = [1,2,3,4,5]
square = []
for i in num:
    i=i**2
    square.append(i)
print(square)
```

USING LIST COMPREHENSION:

```python
num = [1,2,3,4,5]
square = [i**2 for i in num]
print(square)
```

```python
marks = [90,60,45,30]
list=[]

#suppose marks >=90 :"A" , marks >= 60 : "B" , MARKS>=40 : "C" OTHERWISE FAIL
for i in marks:
    if i >= 90:
        list.append("A")
    elif i>=60:
        list.append("B")
    
    elif i>=40:
        list.append("C")
    else:
        list.append("fail")
print(list)
    
    

```

```python
list = ["A" if m>=90 else "B" if m>=60 else "C" if m>=40 else "fail" for m in marks]
print(list)
```

```python
ages = [8,16,35,65]
#categorize ages as below 13:child , below 20: teenage , below 60:adult , and otherwise senior
result = ["Child" if i<13 else "Teenage" if i<20 else "Adult" if i<60 else "Senior" for i in ages]
print(result)
```
