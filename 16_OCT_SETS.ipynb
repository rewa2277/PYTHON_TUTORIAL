SETS METHODS:


 add() : It is used to add single element in a set.

```python
names = {"Priya","Amit","Neha","Rahul"}
names.add("Rewa")
print(names)
```

```python
values = {2,3,4,5}    #add 7 in a set
values.add(7)
print(values)
```

```python
numbers = {10,20,30}
numbers.add(20)
print(numbers)
```

update() : It is used when we want to add more than one element in a set.

```python
numbers = {10,20}
numbers.update({30,40,50,20})
print(numbers)
```

```python
employees = {"Priya" , "Amit" ,"Sakshi" , "Arti" ,"Gaurav"}
#add new employees in the above set : "Chaitali" , "Nikhil" ,"Sneha" ,"Vishakha"
employees.update(["Chaitali" , "Nikhil" ,"Sneha" ,"Vishakha"])
print(employees)
```

```python
a = {1,2,3}
b = {4,5,6}
a.update(b)
print(a)
```

remove() : It is used to remove single element from set

```python
a = {1,2,3}
a.remove(3)
print(a)
```

```python
numbers = {10,20,30,45,67,90}
numbers.remove(45)
```

```python
numbers.remove(100)   # as it is not present it give error
```

```python
numbers = {10,20,30,45,67,90}
numbers.discard(100)
print(numbers)     
```

pop() : Removes an arbitary(random) element

numbers = {10,20,5,30,56,78}
numbers.pop()
print(numbers)

```python
employees = {"Priya" , "Amit" ,"Sakshi" , "Arti" ,"Gaurav"}
employees.pop()
print(employees)
```

clear() : It is used to remove all the element from the set and give empty set

```python
employees.clear()
print(employees)
```

del() : It is used to remove entire existence of that set

```python
del employees
print(employees)
```

```python
salaries = {10000,25000,35000,40000}    
higher_sal =set()
#find salaries greater than 20000
for i in salaries:
    if i>20000:
        higher_sal.add(i)
print(higher_sal)


```
