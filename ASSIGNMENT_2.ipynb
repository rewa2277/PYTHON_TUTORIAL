29. Given text = &quot;Python is powerful and Python is popular&quot;, use lower() and slicing to
extract the first 6 characters and the last 6 characters. Then join the two extracted pieces
using &quot; ... &quot;.

```python
text = "Python is powerful and Python is popular"

text = text.lower()

first_6 = text[:6]
last_6 = text[-6:]

result = first_6 + " ... " + last_6

print(result)
```

30. Given text = &quot; DATA SCIENCE | MACHINE LEARNING | PYTHON PROGRAMMING &quot;,
strip the string, convert it to title case, split it using &quot; | &quot;, take the first and last items, and
join them using &quot; &amp; &quot;.

```python
text = " DATA SCIENCE | MACHINE LEARNING | PYTHON PROGRAMMING "
result ="&".join(text.strip().title().split("|")[::2])
print(result)
```
