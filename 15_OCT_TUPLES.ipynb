zip()

```python
l1 = [10,20,30]   #here we define two variables and make pair of different elements of different list.
l2 = [40,50,60]
result = list(zip(l1,l2))
print(result)
```

```python
addition  = [x+y for x,y in result]   #here we add that different elements 
print(addition)
```

```python
# without zip():
l1 = [10,20,30]   # numbers of element should be same means length should be same 
l2 = [40,50,60]
result = []
for i in range(len(l1)):
    result.append(l1[i]+l2[i])
print(result)
```
