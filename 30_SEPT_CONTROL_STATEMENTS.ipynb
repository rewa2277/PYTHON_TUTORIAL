```python
##Write a program to calculate the category of electricity consumption.
#0–100                  Low
#101–300                Medium
#301–500                High
#Above 500              Very High
```

```python
elect_cons = int(input("Enter your electricity consumption"))
if elect_cons>=0 and elect_cons<=100:
    print("Low")
elif elect_cons>=101 and elect_cons<=300:
    print("Medium")
elif elect_cons>=301 and elect_cons<=500:
    print("High")
else:
    print("Very high")
```
