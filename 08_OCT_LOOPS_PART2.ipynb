NESTED LOOP : IT IS LOOP INSIDE ANOTHER LOOP

```python
for i in range(1,5):
    for j in range(1,5):
        print(i,j)
```

PATTERN USING NESTED LOOPS

```python
n = 5
for i in range(1,5):   #i=number of rows
    for j in range(1,5):  #j = how many times in each line or number of columns
        print("*",end = " ")   #what to print , end = telling that print * next in vertical line 
    print()    # when all values of j is done then move to next line of code
```

```python
for i in range(1,5):
    for j in range(1,i+1):
        print(j,end = " ")
    print()
```

```python
for i in range(1,5):
    for j in range(1,i+1):
        print("*",end = " ")
    print()
```

```python
#take the value from user and print the pattern of numbers starting from 1 to n ( no. of rows) in increasing order
n = int(input("enter your rows:"))
for i in range(n+1):
    for j in range(1,i+1):
        print(j,end = " ")
    print()
    
```

```python
for i in range(1,5):
    for j in range(5-i):
        print("*",end = " ")
    print()
```

```python
for i in range(1,5):
    for j in range(1,5-i+1):
        print(j,end = " ")
    print()
```
