WHILE LOOP : IT EXECUTES UPTO CONDITION TO BECOME TRUE

```python
#syntax:
#while condition:
    #code(write increment parameter) 
```

```python
i = 1 
while i<=5:
    print(i)
    i+=1
print("loop ended")
```

```python
#print even number from 1 to 10 using while loop 
num = 1
while num<=10:
    if num%2==0:
        print(num)
    num+=1
```

```python
#sum of numbers from 1 to 10 using while loop
sum = 0
i = 1
while i <= 10:
    sum+=i
    i+=1
print("sum is:",sum)
```

```python
# print countdown numbers from 10 to 1 using while loop
```

```python
i = 10
while i>=1:   #use smaller 
    print(i)
    i-=1
```

```python
# reverse a number:
```

```python
num = 12345
reverse = 0
while num>0:
    last_digit = num % 10    # 12345%5=5
    reverse = reverse  * 10 + last_digit    #reverse = 0 * 10 + 5 = 5
    num = num//10                           #12345//10 = 12345
print(reverse)
```

```python
#sum of digits from the user
```

```python
num = float(input("enter a number :"))
sum = 0
while num>0:
    last_digit = num% 10
    sum+=last_digit
    num//=10
print(sum)
    
```

BREAK , CONTINUE , AND PASS STATEMENTS

#BREAK : STOPS THE LOOP IMMEDIATELY:

```python
# print numbers from 1 to 10 but stop when number is equal to 5
```

```python
for i in range(11):
    if i == 5:
        break
    print(i)
```

```python
# stop when negative number appears in the list

list = [12,5,3,55,65,23,2,70,-5,8,4,67,45,33,87]
for num in list:
    if num<0:
        break
    print(num)
```

CONTINUE : SKIPS THE CURRENT ITERAION AND EXECUTE THE ANOTHER CODE

```python
for i in range(11):
    if i == 5:
        continue
    print(i)
```

```python
# print odd number from 1 to 10 use the concepts for loop and continue
```

```python
for i in range(11):
    if i%2==0:   #if we want to write odd numbers then we have to skip even number.
        continue
    print(i)
        
```

```python
#print even number from 1 to 10 use the concepts for loop and continue
for i in range(1,11):
    if i % 2!= 0:
        continue
    print(i)
```

PASS STATEMENTS : DO NOTHING FOR NOW . IT ACTS LIKE A PLACEHOLDER

```python
for i in range (1,6):
    if i==3:
        pass
    print(i)
```
