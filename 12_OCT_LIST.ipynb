DIVISIBLE BY 3

```python
numbers = [1,12,24,6,1,71,34,43,89,15,90,81]
divisible_by_3 = []

# make a list of numberes that are divisible by 3

for i in numbers:
    if i% 3 == 0:
        divisible_by_3.append(i)
print(divisible_by_3)
        
```

remove():

```python
students = ["Sakshi","Sukhada","Sampada","Sukhada","Vaishnavi","Ashwini"]
students.remove("Sukhada")
print(students)
```

POP() : POP METHOD REMOVES THE LAST ELEMENT FROM THE LIST.

```python
students = ["chaitali","Rewa","Aditya","Arsh","Nilesh","Ganesh","Kritika","Harshal"]
students.pop()
print(students)
```

```python
students = ["chaitali","Rewa","Aditya","Arsh","Nilesh","Ganesh","Kritika","Harshal"]
students.pop(2)
print(students)
```

sort() : sort () method used to sort elements in ascending order

```python
numbers = [50,45,67,67,23,1]
numbers.sort
print(numbers)
```

descending orders:


```python
numbers = [50,45,67,67,23,1]
numbers.sort(reverse=True)
print(numbers)
```

```python
# sorted() is a built in function:

numbers = [50,45,67,67,23,1]
new_numbers = sorted(numbers)
print(new_numbers)
print(numbers)

```
