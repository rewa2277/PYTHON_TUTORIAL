strip() : It removes leading and trailling spaces that means only first and last spaces

```python
sen = " My name is rewa "
sen.strip()
```

```python
#remove the leading and trailling spaces and give it in title in single line 
sen = " python programming "
sen.strip().title()     # this is called method chaining.
```

lstrip() : remove the space from the left

```python
sen = " python programming "
sen.lstrip()
```

rstrip() : remove the space from right

```python
sen = " python programming     "
sen.rstrip()
```

isalpha() ; checks whether a string purely contain contain alphabets

```python
text = "1234rewa"
text.isalpha()
```

```python
text = "@1234rew"
text.isalpha()
```

isdigit() : checks whether it purely contain digit or not

```python
text = "122345"
text.isdigit()
```

```python
text = "523653hghdjhfm@#"
text.isdigit()
```

alnum() ; checks whether a string contains alphabets or digit

```python
values = "wre23426"
values.isalnum()
```

```python
values = "1426425"
values.isalnum()    # it is or condition so one of the condition is satisfy like here digits are present but alphabets are absent
```

```python
#take a string from user and write whether the string contains only alphabets , only digits ,both of them if nothing and only special characters then print special characters 
```

```python
str = input("write your string here:")
if str.isalpha():
    print("string contains purely alphabets")
elif str.isdigit():
    print("string contains purely digits")
elif str.isalnum():
    print("string contains both alphabets or digits")
else:
    print("string contains special characters")
```
